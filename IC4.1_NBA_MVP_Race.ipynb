# ----SPECIFIC FOR PREP/IC/HW ----

```python
%%sql
create table NBA_MVPLadder(
  player_name text,
  player_team text,
  player_ppg real,
  MVP_rank integer
) Strict;
```

```python
%%sql
insert into NBA_MVPLadder (player_name, player_team, player_ppg, MVP_rank)
values
  ('Nikola Jokic', 'Denver Nuggets', 28.7, 2),
  ('Shai Gilgeous-Alexander', 'Oklahoma City Thunder', 31.8, 1),
  ('James Harden', 'Los Angeles Clippers', 25.0, 12),
  ('Luka Doncic', 'Los Angeles Lakers', 32.8, 3);
```

```python
%%sql
select * from NBA_MVPLadder
where player_ppg > 30.0
```

```python
%%sql
update NBA_MVPLadder
set player_team = 'Cleveland Cavaliers'
where player_name = 'James Harden';

delete from NBA_MVPLadder
where player_name = 'Luka Doncic';
```

```python
#Final Updated Table
%%sql
select * from NBA_MVPLadder;
```
