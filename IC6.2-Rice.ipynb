2. Speed to Insight: How does the architecture of a "Leader Node" and "Compute Nodes" change how quickly a Business Intelligence report is generated?  https://docs.aws.amazon.com/redshift/latest/dg/c_high_level_system_architecture.html


## YOUR ANSWERS TO PART 2 BELOW

Question 1: Columnar storage is more important for an analyst because instead of reading every row, Redshift only reads the columns needed for a query. This reduces the amount of data that needs to be loaded into memory and improves compression which saves disk space and speeds up the time it takes an analyst to get results.

Question 2: The leader node receives a query request and develops code for an execution plan that contains the necessary steps for the desired results. It then distributes this code to compute nodes which execute simultaneously on different chunks of data, and then send the results back to the leader node for final aggregation. By distrubuting the work load among multiple nodes, this archetecture allows for a BI report to be generated much faster.
